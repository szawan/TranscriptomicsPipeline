```python
import pandas as pd

df = pd.read_csv("/scratch/sah2p/datasets/2023_11_04_BurkeLab/output/9_DEG_new/01_FC2_DEGenesList.csv", sep="\t")
```

```python
df.columns
```

```python
cols = ['H1975_E07_vs_H1975_C36','H1975_C36_vs_H1975_Veh','H1975_E07_vs_H1975_Veh'] 
li_genes = []
for col in cols:
    df[col] = df[col].str.replace("gene-", "")
    li_genes += df[~df[col].isna()][col].tolist()

print(li_genes)
print(len(li_genes))
print(len(set(li_genes)))

li_genes = list(set(li_genes))

print(li_genes)
```

```python
#Write to file

file_path = '/scratch/sah2p/datasets/2023_11_04_BurkeLab/output/9_DEG_new/A549_gene_set.txt'

# Open the file in write mode
with open(file_path, 'w') as file:
    # Convert each element of the list to a string and write it to the file
    for item in li_genes:
        file.write(str(item) + '\n')
```
